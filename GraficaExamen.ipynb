```python
import numpy as np
import matplotlib.pyplot as plt

# Complejidades exactas aproximadas del algoritmo COMPLETO (ordenamiento + voraz)
# Tomamos la parte de ordenamiento como n·log2(n) y la parte voraz como lineal
def t_mejor(n):
    return n * np.log2(n) + n          # mejor: T(n) ≈ n log n + n

def t_medio(n):
    return n * np.log2(n) + (2*n - 1)  # medio: T(n) ≈ n log n + (2n - 1)

def t_peor(n):
    return n * np.log2(n) + (3*n - 2)  # peor: T(n) ≈ n log n + (3n - 2)

# Rango de tamaños de entrada
n = np.arange(2, 201)   # desde 2 para evitar log(1) = 0

# Evaluación de las funciones
y_mejor = t_mejor(n)
y_medio = t_medio(n)
y_peor  = t_peor(n)

# Gráfica
plt.figure(figsize=(10, 6))
plt.plot(n, y_mejor, label="Mejor caso  T(n) ≈ n·log₂n + n")
plt.plot(n, y_medio, label="Caso medio T(n) ≈ n·log₂n + 2n - 1")
plt.plot(n, y_peor,  label="Peor caso  T(n) ≈ n·log₂n + 3n - 2")

plt.title("Comparación de la complejidad del algoritmo de asesorías\n(mejor, medio y peor caso)")
plt.xlabel("n (número de solicitudes)")
plt.ylabel("T(n) (operaciones aproximadas)")
plt.grid(True)
plt.legend()
plt.show()
```
